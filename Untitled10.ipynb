```python
# ─── CELL 1: Install & Imports ───────────────────────────────────────────────

import os, random, numpy as np, warnings
warnings.filterwarnings('ignore')

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

print(" TensorFlow:", tf.__version__)
print(" GPU:", tf.config.list_physical_devices('GPU'))
```

```python
# ─── CELL 2: Mount Google Drive & Set Dataset Path ───

from google.colab import drive
drive.mount('/content/drive')

# ⌄  CHANGE THIS to your actual folder path in Drive
DATASET_PATH = '/content/drive/MyDrive/DDD_FINAL'

# Verify
for cls in os.listdir(DATASET_PATH):
    p = os.path.join(DATASET_PATH, cls)
    if os.path.isdir(p):
        print(f"၄ {cls}: {len(os.listdir(p))} images")
```

```python
# ─── CELL 3: Configuration ───────────────────────────────────────────────────

IMG_HEIGHT       = 224
IMG_WIDTH        = 224
BATCH_SIZE       = 32
EPOCHS_PHASE1    = 30
EPOCHS_PHASE2    = 20
LEARNING_RATE    = 1e-4
FINE_TUNE_LR     = 1e-5
VALIDATION_SPLIT = 0.15
SEED             = 42
CLASS_NAMES      = ['Drowsy-samples (1)', 'Non Drowsy-samples (1)']

print("✅ Config ready")
```

```python
# ─── CELL 4: Visualise Sample Images ─────────────────────────────────────────

fig, axes = plt.subplots(2, 5, figsize=(16, 7))
fig.suptitle('Sample Dataset Images', fontsize=15, fontweight='bold')

for row, cls in enumerate(CLASS_NAMES):
    cls_path  = os.path.join(DATASET_PATH, cls)
    samples   = random.sample(os.listdir(cls_path), 5)
    for col, fname in enumerate(samples):
        img = load_img(os.path.join(cls_path, fname),
                       target_size=(IMG_HEIGHT, IMG_WIDTH))
        axes[row][col].imshow(img)
        axes[row][col].set_title(cls.upper(), fontsize=9,
                                 color='red' if 'Drowsy' in cls else 'green')
        axes[row][col].axis('off')

plt.tight_layout()
plt.show()
```

```python
# ─── CELL 6: Build Model (Transfer Learning) ───

base_model = MobileNetV2(weights='imagenet', include_top=False,
                         input_shape=(IMG_HEIGHT, IMG_WIDTH, 3))

base_model.trainable = False

model = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(128, activation='relu'),
    layers.BatchNormalization(),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=LEARNING_RATE),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("\n✅ Model architecture built and compiled.")
```

```python
# ─── CELL 5: Data Generators & Augmentation ───

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.15,
    zoom_range=0.15,
    horizontal_flip=True,
    fill_mode='nearest',
    validation_split=VALIDATION_SPLIT
)

val_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=VALIDATION_SPLIT
)

train_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='binary',
    shuffle=True,
    seed=SEED,
    subset='training'
)

validation_generator = val_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='binary',
    shuffle=False,
    seed=SEED,
    subset='validation'
)

print(f"\n✅ Generators initialized: {train_generator.samples} train / {validation_generator.samples} validation images")
```
